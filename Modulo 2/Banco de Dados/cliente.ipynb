```python
cliente = 'clientes.db'
```

```python
import sqlite3

script_clientes = ''' 
CREATE TABLE IF NOT EXISTS Clientes(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nome TEXT NOT NULL,
    idade INTEGER NOT NULL,
    email TEXT NOT NULL
);'''

try:
    with sqlite3.connect("cliente.db") as con:
        cur = con.cursor()
        cur.execute(script_clientes)
        con.commit()
        print("Tabela criada com sucesso!")
        
except sqlite3.OperationalError as e:
    print('Erro:', e)
```

```python
sql = 'SELECT * FROM Clientes'

try:
    with sqlite3.connect(cliente) as con:
        cur = con.cursor()

        cur.execute(sql)

        res = cur.fetchall()

        print(res)

except sqlite3.OperationalError as e:
    print('Erro:', e)
```

```python
sql = 'INSERT INTO Clientes (nome, idade, email) VALUES (?, ?, ?)'
try:
    with sqlite3.connect(cliente) as con:
        cur = con.cursor()
        cur.execute(sql, ('Felipe', 15, 'felipe3@gmail.com'))
        con.commit()
except sqlite3.OperationalError as e:
    print('Erro:', e)
```

```python
sql = 'UPDATE Clientes SET idade = ? WHERE id = ?'

try:
    with sqlite3.connect(cliente) as con:
        cur = con.cursor()
        cur.execute(sql,(28, 1))
        con.commit()
except sqlite3.OperationalError as e:
    print('Erro', e)
```

```python
sql = 'DELETE FROM Clientes WHERE id = ?'
try:
    with sqlite3.connect(cliente) as con:
        cur = con.cursor()
        cur.execute(sql,(2,))
        con.commit()
except sqlite3.OperationalError as e:
    print('Erro', e)
```

```python
con.close()
print("Conexão encerrada.")
```
